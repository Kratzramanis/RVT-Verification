```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# RVT-Feldeffekt & Trichterprojektion: NGC 253 + NGC 247
# Bezugspunkt: Erde im Ursprung (0,0) zur Wahrung der Symmetrie
# ============================================================

# --- Parameter ---
N = 140
L = 40.0         # Systemgröße in Millionen Lichtjahren (Mio. Lj)
C_RVT = 0.8

# Zielausrichtung der Sichtachsen an der Erde: Exakt 90°
target_angle = np.radians(90.0)
gamma_90 = 1.0 + C_RVT * np.sin(target_angle)

# --- Realer Wertekanon (Massen in M_sun, Distanzen in Mio. Lj) ---
# 1. NGC 253 (Sculptor-Galaxie) - Edge-on Starburst
M_253 = 5.0e10
scale_253 = 2.5   # Effektive RVT-Feld-Skalenlänge
center_253 = (11.5, 0.0, 0.0) # Liegt exakt auf der X-Achse

# 2. NGC 247 (Begleitgalaxie) - Zwergspirale
M_247 = 3.0e9
scale_247 = 1.2   # Kleinere Skalenlänge entsprechend der Masse
center_247 = (0.0, 11.1, 0.0) # Liegt exakt auf der Y-Achse (90° zur X-Achse)

# Reale Inklinationen (für die Amplitudenmodifikation)
inclination_253 = np.radians(78.0)  # Nahezu Edge-on
inclination_247 = np.radians(74.0)  # Ebenfalls hohe Inklination

gamma_253 = 1.0 + C_RVT * np.sin(inclination_253)
gamma_247 = 1.0 + C_RVT * np.sin(inclination_247)

print(f"RVT-Kopplung an der Erde (90°) = {gamma_90:.3f}")
print(f"gamma_RVT(NGC 253)            = {gamma_253:.3f}")
print(f"gamma_RVT(NGC 247)            = {gamma_247:.3f}")

# --- Gitter erstellen (3D-Basis, Erde liegt bei 0,0,0) ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]

X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Gaußförmige Basisfelder ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

# Inklinationsberechnung der Felder
Xi_253_incl = gaussian_field(M_253, scale_253, center_253) * gamma_253
Xi_247_incl = gaussian_field(M_247, scale_247, center_247) * gamma_247
Xi = Xi_253_incl + Xi_247_incl

# --- Gradienten und Ableitungen für den Hintergrund ---
gx, gy, gz = np.gradient(Xi, dx, dx, dx, edge_order=2)
V_12 = 2.0 * C_RVT * (gx**2 + gy**2 + gz**2) # Autokorrelierte Feldeffekt-Varianz

# --- Extraktion des 2D-Schnitts bei z = 0 ---
z_slice = N // 2
X2 = X[:, :, z_slice]
Y2 = Y[:, :, z_slice]
V_12_slice = V_12[:, :, z_slice]

# --- Trichterfunktion (Geöffnet zwischen den orthogonalen Achsen) ---
def orthogonal_funnel_2d(X2, Y2, c253, c247, mass_253, mass_247, scale_253, scale_247):
    # Relativkoordinaten zu den Galaxienzentren
    X_rel_253 = X2 - c253[0]
    Y_rel_253 = Y2 - c253[1]

    X_rel_247 = X2 - c247[0]
    Y_rel_247 = Y2 - c247[1]

    # Trichteröffnungen steuern (Trichter öffnen sich zum Interaktionsraum hin)
    # NGC 253 öffnet sich primär entlang der Y-Richtung zur Mitte hin
    funnel_253 = - (mass_253 / 1e10) * np.exp(-(Y_rel_253**2) / (2.0 * (scale_253 * (1.0 + np.abs(X_rel_253)/2.0))**2))

    # NGC 247 öffnet sich primär entlang der X-Richtung zur Mitte hin
    funnel_247 = - (mass_247 / 1e10) * np.exp(-(X_rel_247**2) / (2.0 * (scale_247 * (1.0 + np.abs(Y_rel_247)/2.0))**2))

    return funnel_253 + funnel_247

# Trichter generieren
funnel_total = orthogonal_funnel_2d(X2, Y2, center_253, center_247, M_253, M_247, scale_253, scale_247)

# Normierung für sauberes Blending
def normalize(A):
    return A / (np.max(np.abs(A)) + 1e-30)

projection_map = normalize(funnel_total) + 0.4 * normalize(V_12_slice)

# --- Visualisierung der 2D-Trichterkarte ---
plt.figure(figsize=(12, 10))

plt.imshow(
    projection_map.T,
    origin='lower',
    extent=[-L / 2, L / 2, -L / 2, L / 2],
    cmap='coolwarm',
    aspect='equal'
)

# Konturlinien für den Krümmungsverlauf des 2D-Äthers
plt.contour(X2, Y2, funnel_total, levels=14, colors='black', alpha=0.35, linewidths=0.8)

# Die physischen Objekte einzeichnen (Erde im Nullpunkt!)
plt.scatter(0, 0, c='blue', s=250, marker='o', edgecolors='white', label='Erde (RVT-Nullpunkt / 90°)')
plt.scatter(center_253[0], center_253[1], c='cyan', s=180, marker='x', linewidths=3, label='NGC 253 (Starburst)')
plt.scatter(center_247[0], center_247[1], c='orange', s=100, marker='x', linewidths=3, label='NGC 247 (Zwergspirale)')

# Unendlich durchgezogene Sichtachsen (Katheten) einzeichnen
plt.axhline(0, color='cyan', linestyle='--', alpha=0.5, linewidth=1.5, label='Sichtachse A (X-Achse)')
plt.axvline(0, color='orange', linestyle='--', alpha=0.5, linewidth=1.5, label='Sichtachse B (Y-Achse)')

# Hypotenuse (Der 45°-Projektionskorridor)
plt.plot([center_253[0], center_247[0]], [center_253[1], center_247[1]], color='red', linestyle=':', linewidth=2.5, label='Hypotenuse (45°-Ätherlinie)')

# Beschriftungen und Texte
plt.text(2, 2, 'Erde', color='white', fontweight='bold', fontsize=12)
plt.text(center_253[0], center_253[1] + 1.5, 'NGC 253', color='cyan', fontweight='bold', ha='center')
plt.text(center_247[0] + 1.5, center_247[1], 'NGC 247', color='orange', fontweight='bold', va='center')

plt.colorbar(label='Normierte RVT-Projektionsspannung')
plt.title('RVT-Trichterprojektion: NGC 253 & NGC 247\nBezugssystem Erde im absoluten Nullpunkt (0,0)', fontsize=14, fontweight='bold')
plt.xlabel('X-Achse (Mio. Lichtjahre)')
plt.ylabel('Y-Achse (Mio. Lichtjahre)')
plt.xlim([-5, L/2])
plt.ylim([-5, L/2])
plt.grid(True, alpha=0.2)
plt.legend(loc='upper right')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ============================================================
# ERWEITERTE COUPLING-MATRIX: RVT-Tensorfelder für NGC 253 + NGC 247
# Bezugssystem: Erde im absoluten Nullpunkt (0,0,0)
# ============================================================

# --- Parameter ---
N = 140          # Gitterpunkte für drakonische Präzision
L = 40.0         # Systemgröße in Millionen Lichtjahren (Fenster: [-20, +20] Mio. Lj)
C_RVT = 0.8      # RVT-Kopplungskonstante

# --- Realer Wertekanon (Mio. Lj & M_sun) ---
# NGC 253: Massereicher Starburst auf der X-Achse
M_253 = 5.0e10
scale_253 = 2.5
center_253 = (11.5, 0.0, 0.0)
inclination_253_deg = 78.0

# NGC 247: Masseärmerer Partner auf der Y-Achse (Exakt 90° an der Erde)
M_247 = 3.0e9
scale_247 = 1.2
center_247 = (0.0, 11.1, 0.0)
inclination_247_deg = 74.0

# Winkel in Radiant umrechnen
inclination_253 = np.radians(inclination_253_deg)
inclination_247 = np.radians(inclination_247_deg)

# --- RVT-Gamma-Korrektur ---
gamma_253 = 1.0 + C_RVT * np.sin(inclination_253)
gamma_247 = 1.0 + C_RVT * np.sin(inclination_247)

print(f"gamma_RVT(NGC 253, {inclination_253_deg:.0f}°) = {gamma_253:.3f}")
print(f"gamma_RVT(NGC 247, {inclination_247_deg:.0f}°) = {gamma_247:.3f}")

# --- Gitter erstellen ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]

X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Gaußförmige Projektionsfelder mit Inklinationsskalierung \sin(i) ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_1 = gaussian_field(M_253, scale_253, center_253) * np.sin(inclination_253)
Xi_2 = gaussian_field(M_247, scale_247, center_247) * np.sin(inclination_247)
Xi = Xi_1 + Xi_2

# --- Gradienten berechnen ---
def gradient_3d(f, dx):
    gx, gy, gz = np.gradient(f, dx, dx, dx, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

grad_Xi_1 = gradient_3d(Xi_1, dx)
grad_Xi_2 = gradient_3d(Xi_2, dx)

# --- Feldeffekt-Varianz V_12 ---
V_12 = 2.0 * C_RVT * np.sum(grad_Xi_1 * grad_Xi_2, axis=-1)

# --- Zweite Ableitungen / Laplace ---
def second_derivative(f, dx, axis):
    first = np.gradient(f, dx, axis=axis, edge_order=2)
    second = np.gradient(first, dx, axis=axis, edge_order=2)
    return second

d2x_Xi = second_derivative(Xi, dx, axis=0)
d2y_Xi = second_derivative(Xi, dx, axis=1)
d2z_Xi = second_derivative(Xi, dx, axis=2)

laplace_Xi = d2x_Xi + d2y_Xi + d2z_Xi

# --- RVT-Korrekturtensor E_mu_nu ---
E_xx = d2x_Xi - laplace_Xi
E_yy = d2y_Xi - laplace_Xi
E_zz = d2z_Xi - laplace_Xi

# --- Projektion auf die 2D-Ätherebene ---
S_2D = E_xx + E_yy
S_perp = E_zz
M_RVT = S_2D - S_perp

# --- 2D-Schnitt im Detektorzentrum bei z = 0 ---
z_slice = N // 2
V_12_slice = V_12[:, :, z_slice]
M_RVT_slice = M_RVT[:, :, z_slice]
S_2D_slice = S_2D[:, :, z_slice]

# --- Visualisierung der drei tensoriellen Feldkomponenten ---
plt.figure(figsize=(18, 5))

# Plot 1: Feldeffekt-Varianz (Kreuzterm der orthogonalen Kraftlinien)
plt.subplot(1, 3, 1)
plt.imshow(V_12_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='viridis', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['blue', 'cyan', 'orange'], s=60, marker='o', edgecolors='black')
plt.colorbar(label=r'$\mathcal{V}_{12}$')
plt.title(r'Feldeffekt-Varianz $\mathcal{V}_{12}$' + '\nNGC 253 + NGC 247')
plt.xlabel('x [Mio. Lj]')
plt.ylabel('y [Mio. Lj]')

# Plot 2: 2D-Projektionsspannung (Die Geometrie der Null-Grad-Ebene)
plt.subplot(1, 3, 2)
plt.imshow(S_2D_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='plasma', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['blue', 'cyan', 'orange'], s=60, marker='o', edgecolors='black')
plt.colorbar(label=r'$\mathcal{S}_{2D}$')
plt.title(r'2D-Projektionsspannung $\mathcal{S}_{2D}$')
plt.xlabel('x [Mio. Lj]')
plt.ylabel('y [Mio. Lj]')

# Plot 3: RVT-Speicherfunktion (Der haptische Energierückstand)
plt.subplot(1, 3, 3)
plt.imshow(M_RVT_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='coolwarm', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['blue', 'cyan', 'orange'], s=60, marker='o', edgecolors='black')
plt.colorbar(label=r'$\mathcal{M}_{\mathrm{RVT}}$')
plt.title(r'RVT-Speicherfunktion $\mathcal{M}_{\mathrm{RVT}}$')
plt.xlabel('x [Mio. Lj]')
plt.ylabel('y [Mio. Lj]')

plt.tight_layout()
plt.show()

# --- 3D-Schnitt-Visualisierung ---
fig = plt.figure(figsize=(12, 8))
ax = fig.add_subplot(111, projection='3d')

for z_fix in [N // 4, N // 2, 3 * N // 4]:
    ax.contour(X[:, :, z_fix], Y[:, :, z_fix], V_12[:, :, z_fix], levels=10, cmap='viridis', alpha=0.4, zdir='z', offset=z[z_fix])

ax.scatter(0, 0, 0, color='blue', s=100, marker='o', label='Erde (Nullpunkt)')
ax.scatter(center_253[0], center_253[1], center_253[2], color='cyan', s=80, marker='x', label='NGC 253')
ax.scatter(center_247[0], center_247[1], center_247[2], color='orange', s=80, marker='x', label='NGC 247')

ax.set_xlim(-L / 2, L / 2)
ax.set_ylim(-L / 2, L / 2)
ax.set_zlim(-L / 2, L / 2)
ax.set_xlabel('x [Mio. Lj]')
ax.set_ylabel('y [Mio. Lj]')
ax.set_zlabel('z [Mio. Lj]')
ax.set_title(r'3D-Feldeffekt-Varianz $\mathcal{V}_{12}$ (Erde als Koordinatenursprung)')
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# TRILATERALE RVT-MATRIZEN: Paarweiser Winkelabgleich (Kosinus)
# Sgr A* (0,0,0) [Masse: 4.154e6 M_sun]
# NGC 253 (11.5, 0, 0) & NGC 247 (0, 11.1, 0)
# ============================================================

N = 140
L = 40.0         # Fenster: [-20, +20] Mio. Lj
C_RVT = 0.8

# --- Reale Quellendaten ---
M_SgrA = 4.154e6;  scale_SgrA = 0.5;  c_SgrA = np.array([0.0, 0.0, 0.0])
M_253  = 5.0e10;   scale_253  = 2.5;  c_253  = np.array([11.5, 0.0, 0.0])
M_247  = 3.0e9;    scale_247  = 1.2;  c_247  = np.array([0.0, 11.1, 0.0])

# --- Gitter (2D-Schnitt bei z=0 direkt berechnet für Speed) ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
X, Y = np.meshgrid(x, y, indexing='ij')

# --- Funktion für Richtungsvektoren und Distanzen zu jedem Punkt ---
def get_vector_fields(X, Y, center):
    rx = X - center[0]
    ry = Y - center[1]
    dist = np.sqrt(rx**2 + ry**2 + 1e-9)
    return rx / dist, ry / dist, dist

# Einheitsvektoren von jedem Gitterpunkt zu den 3 Zentren
ux_SgrA, uy_SgrA, d_SgrA = get_vector_fields(X, Y, c_SgrA)
ux_253,  uy_253,  d_253  = get_vector_fields(X, Y, c_253)
ux_247,  uy_247,  d_247  = get_vector_fields(X, Y, c_247)

# --- Gaußsche Basisfelder ---
Xi_SgrA = M_SgrA * np.exp(-(d_SgrA**2) / (2 * scale_SgrA**2))
Xi_253  = M_253  * np.exp(-(d_253**2)  / (2 * scale_253**2))
Xi_247  = M_247  * np.exp(-(d_247**2)  / (2 * scale_247**2))

# --- Gradienten der Einzelfelder (skalarer Betrag für die Kopplung) ---
def get_scalar_grad(Xi, dist, scale):
    # Analytischer Gauß-Gradientenbetrag: dXi/dr = -r/scale^2 * Xi
    return (dist / (scale**2)) * Xi

g_SgrA = get_scalar_grad(Xi_SgrA, d_SgrA, scale_SgrA)
g_253  = get_scalar_grad(Xi_253, d_253, scale_253)
g_247  = get_scalar_grad(Xi_247, d_247, scale_247)

# --- Paarweiser Richtungsabgleich (Cosinus der echten Geometrie-Winkel) ---
# 1. Abgleich: Sgr A* zu NGC 253
cos_theta_1 = ux_SgrA * ux_253 + uy_SgrA * uy_253
V_SgrA_253 = g_SgrA * g_253 * cos_theta_1

# 2. Abgleich: Sgr A* zu NGC 247
cos_theta_2 = ux_SgrA * ux_247 + uy_SgrA * uy_247
V_SgrA_247 = g_SgrA * g_247 * cos_theta_2

# 3. Abgleich: NGC 253 zu NGC 247 (Die Hypotenusen-Interaktion)
cos_theta_3 = ux_253 * ux_247 + uy_253 * uy_247
V_253_247 = g_253 * g_247 * cos_theta_3

# --- Totale trilaterale Speicherfunktion M_RVT ---
# Fusioniert die echten physischen Dreiecksspannungen
M_RVT_trilateral = C_RVT * (V_SgrA_253 + V_SgrA_247 + V_253_247)

# --- Trichterfunktion mit integriertem Sgr A* Peak ---
funnel_253 = - (M_253 / 1e10) * np.exp(-((Y - c_253[1])**2) / (2.0 * (scale_253 * (1.0 + np.abs(X - c_253[0])/2.0))**2))
funnel_247 = - (M_247 / 1e10) * np.exp(-((X - c_247[0])**2) / (2.0 * (scale_247 * (1.0 + np.abs(Y - c_247[1])/2.0))**2))
funnel_SgrA = - (M_SgrA / 1e6) * np.exp(-(d_SgrA**2) / (2.0 * scale_SgrA**2))
funnel_total = funnel_253 + funnel_247 + funnel_SgrA

def normalize(A):
    return A / (np.max(np.abs(A)) + 1e-30)

projection_map = normalize(funnel_total) + 0.5 * normalize(M_RVT_trilateral)

# --- VISUALISIERUNG ---
plt.figure(figsize=(18, 7))

# Plot 1: Die trilaterale Speicherfunktion (Echter Kosinus-Winkelabgleich)
plt.subplot(1, 2, 1)
plt.imshow(M_RVT_trilateral.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='coolwarm', aspect='equal')
plt.scatter(c_SgrA[0], c_SgrA[1], c='magenta', s=200, marker='o', edgecolors='white', label='Sgr A* (Masse-Anker)')
plt.scatter(c_253[0], c_253[1], c='cyan', s=150, marker='x', linewidths=3, label='NGC 253')
plt.scatter(c_247[0], c_247[1], c='orange', s=120, marker='x', linewidths=3, label='NGC 247')
# Dreiecks-Symmetrielinien einzeichnen
plt.plot([0, c_253[0]], [0, c_253[1]], 'white', linestyle='--', alpha=0.6)
plt.plot([0, c_247[0]], [0, c_247[1]], 'white', linestyle='--', alpha=0.6)
plt.plot([c_253[0], c_247[0]], [c_253[1], c_247[1]], 'red', linestyle=':', linewidth=2, label='Asymmetrische Hypotenuse')
plt.colorbar(label=r'Trilaterale Speicherwirkung $\mathcal{M}_{\mathrm{RVT}}$')
plt.title(r'Trilaterale Speicherfunktion $\mathcal{M}_{\mathrm{RVT}}$ (Paarweiser $\cos\theta$-Abgleich)', fontsize=12, fontweight='bold')
plt.xlabel('X-Achse (Mio. Lj)'); plt.ylabel('Y-Achse (Mio. Lj)')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])
plt.legend()

# Plot 2: Kombinierte Projektionskarte (Trichter + Trilaterale Spannung)
plt.subplot(1, 2, 2)
plt.imshow(projection_map.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='plasma', aspect='equal')
plt.contour(X, Y, funnel_total, levels=14, colors='black', alpha=0.3)
plt.scatter(c_SgrA[0], c_SgrA[1], c='magenta', s=200, edgecolors='white')
plt.scatter(c_253[0], c_253[1], c='cyan', s=150, marker='x', linewidths=3)
plt.scatter(c_247[0], c_247[1], c='orange', s=120, marker='x', linewidths=3)
plt.title('Gesamtes RVT-Projektionsfeld (Trichter & Kosinus-Spannung)', fontsize=12, fontweight='bold')
plt.xlabel('X-Achse (Mio. Lj)')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ============================================================
# TRILATERALE COUPLING-MATRIX (SINUS-MODUS)
# Bezugssystem: Sgr A* im Koordinatenursprung (0,0,0) mit Masse!
# ============================================================

# --- Parameter ---
N = 140          # Gitterpunkte für drakonische Präzision
L = 40.0         # Systemgröße in Millionen Lichtjahren (Fenster: [-20, +20] Mio. Lj)
C_RVT = 0.8      # RVT-Kopplungskonstante

# --- Realer Wertekanon (Mio. Lj & M_sun) ---
# 1. Sgr A* (Zentraler Anker im Nullpunkt)
M_SgrA = 4.154e6
scale_SgrA = 0.5
center_SgrA = (0.0, 0.0, 0.0)

# 2. NGC 253 (Massereicher Starburst auf der X-Achse)
M_253 = 5.0e10
scale_253 = 2.5
center_253 = (11.5, 0.0, 0.0)
inclination_253_deg = 78.0

# 3. NGC 247 (Partner auf der Y-Achse)
M_247 = 3.0e9
scale_247 = 1.2
center_247 = (0.0, 11.1, 0.0)
inclination_247_deg = 74.0

# Winkel in Radiant umrechnen
inclination_253 = np.radians(inclination_253_deg)
inclination_247 = np.radians(inclination_247_deg)

# --- RVT-Gamma-Korrektur (Sgr A* erhält lokalen 90°-Sichtachsenanker)
gamma_SgrA = 1.0 + C_RVT * np.sin(np.radians(90.0))
gamma_253 = 1.0 + C_RVT * np.sin(inclination_253)
gamma_247 = 1.0 + C_RVT * np.sin(inclination_247)

print(f"gamma_RVT(Sgr A*)   = {gamma_SgrA:.3f}")
print(f"gamma_RVT(NGC 253)  = {gamma_253:.3f}")
print(f"gamma_RVT(NGC 247)  = {gamma_247:.3f}")

# --- Gitter erstellen ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]

X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Generierung der 3 eigenständigen Gauß-Projektionsfelder ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_SgrA = gaussian_field(M_SgrA, scale_SgrA, center_SgrA) * gamma_SgrA
Xi_253  = gaussian_field(M_253, scale_253, center_253) * np.sin(inclination_253)
Xi_247  = gaussian_field(M_247, scale_247, center_247) * np.sin(inclination_247)

# Gesamtätherfeld Xi
Xi = Xi_SgrA + Xi_253 + Xi_247

# --- 3D-Gradienten der einzelnen Felder für den Sinus-Abgleich ---
def gradient_3d(f, dx):
    gx, gy, gz = np.gradient(f, dx, dx, dx, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

grad_SgrA = gradient_3d(Xi_SgrA, dx)
grad_253  = gradient_3d(Xi_253, dx)
grad_247  = gradient_3d(Xi_247, dx)

# --- Paarweise Feldeffekt-Varianzen über die Kreuzkomponenten (Sinus-Modulation) ---
# Da sin(theta) = |A x B| / (|A||B|), nutzen wir den orthogonalen komplementären Betrag
def cross_variance_sin(g1, g2):
    # Beträge der Gradientenvektoren
    mag1 = np.linalg.norm(g1, axis=-1) + 1e-30
    mag2 = np.linalg.norm(g2, axis=-1) + 1e-30
    # Skalarprodukt für Cosinus
    dot = np.sum(g1 * g2, axis=-1)
    cos_theta = dot / (mag1 * mag2)
    # Sinus über die trigonometrische Identität bestimmen
    sin_theta = np.sqrt(1.0 - np.clip(cos_theta**2, 0.0, 1.0))
    return 2.0 * C_RVT * mag1 * mag2 * sin_theta

V_SgrA_253 = cross_variance_sin(grad_SgrA, grad_253)
V_SgrA_247 = cross_variance_sin(grad_SgrA, grad_247)
V_253_247  = cross_variance_sin(grad_253, grad_247)

# Totale trilateral verschränkte Varianz
V_12_total = V_SgrA_253 + V_SgrA_247 + V_253_247

# --- Zweite Ableitungen / Laplace für den Gesamttensor ---
def second_derivative(f, dx, axis):
    first = np.gradient(f, dx, axis=axis, edge_order=2)
    return np.gradient(first, dx, axis=axis, edge_order=2)

d2x_Xi = second_derivative(Xi, dx, axis=0)
d2y_Xi = second_derivative(Xi, dx, axis=1)
d2z_Xi = second_derivative(Xi, dx, axis=2)

laplace_Xi = d2x_Xi + d2y_Xi + d2z_Xi

# --- RVT-Korrekturtensor E_mu_nu ---
E_xx = d2x_Xi - laplace_Xi
E_yy = d2y_Xi - laplace_Xi
E_zz = d2z_Xi - laplace_Xi

# --- Projektion auf die 2D-Ätherebene ---
S_2D = E_xx + E_yy
S_perp = E_zz
M_RVT = S_2D - S_perp

# --- Extraction des 2D-Schnitts bei z = 0 ---
z_slice = N // 2
V_12_slice = V_12_total[:, :, z_slice]
M_RVT_slice = M_RVT[:, :, z_slice]
S_2D_slice = S_2D[:, :, z_slice]

# --- Visualisierung der drei tensoriellen Feldkomponenten ---
plt.figure(figsize=(18, 5))

# Plot 1: Trilaterale Feldeffekt-Varianz (Sinus-Kopplung)
plt.subplot(1, 3, 1)
plt.imshow(V_12_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='viridis', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=70, edgecolors='black')
plt.colorbar(label=r'$\mathcal{V}_{12\mathrm{\,(trilateral)}}$')
plt.title(r'Trilaterale Sinus-Varianz $\mathcal{V}_{12}$')
plt.xlabel('x [Mio. Lj]'); plt.ylabel('y [Mio. Lj]')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

# Plot 2: 2D-Projektionsspannung
plt.subplot(1, 3, 2)
plt.imshow(S_2D_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='plasma', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=70, edgecolors='black')
plt.colorbar(label=r'$\mathcal{S}_{2D}$')
plt.title(r'2D-Projektionsspannung $\mathcal{S}_{2D}$')
plt.xlabel('x [Mio. Lj]')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

# Plot 3: RVT-Speicherfunktion (Zentraler Massenanker aktiv!)
plt.subplot(1, 3, 3)
plt.imshow(M_RVT_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='coolwarm', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=70, edgecolors='black')
plt.colorbar(label=r'$\mathcal{M}_{\mathrm{RVT}}$')
plt.title(r'RVT-Speicherfunktion $\mathcal{M}_{\mathrm{RVT}}$')
plt.xlabel('x [Mio. Lj]')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

plt.tight_layout()
plt.show()

# --- 3D-Schnitt-Visualisierung ---
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
for z_fix in [N // 4, N // 2, 3 * N // 4]:
    ax.contour(X[:, :, z_fix], Y[:, :, z_fix], V_12_total[:, :, z_fix], levels=10, cmap='viridis', alpha=0.4, zdir='z', offset=z[z_fix])

ax.scatter(0, 0, 0, color='magenta', s=120, marker='o', label='Sgr A* (4.154 Mio. M_sun)')
ax.scatter(center_253[0], center_253[1], center_253[2], color='cyan', s=80, marker='x', label='NGC 253')
ax.scatter(center_247[0], center_247[1], center_247[2], color='orange', s=80, marker='x', label='NGC 247')
ax.set_xlim(-5, L/2); ax.set_ylim(-5, L/2); ax.set_zlim(-L/4, L/4)
ax.set_xlabel('x [Mio. Lj]'); ax.set_ylabel('y [Mio. Lj]')
ax.set_title(r'3D-Trilaterale Varianz (Sgr A* Masse-Kompensation im Ursprung)')
ax.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# TRILATERALE 9-PUNKT-KOPPLUNGSMATRIX MIT INVERSER DISTANZ
# Bezugssystem: Sgr A* (0,0,0), NGC 253 (11.5, 0, 0), NGC 247 (0, 11.1, 0)
# Toleranzkorrektur: 26.000 Lj (0.026 Mio. Lj)
# ============================================================

# --- Parameter ---
N = 140
L = 40.0
C_RVT = 0.8
TOLERANZ = 0.026  # 26.000 Lichtjahre Toleranzgrenze in Mio. Lj

# --- Massen und Positionen ---
M_SgrA = 4.154e6;  c_SgrA = np.array([0.0, 0.0, 0.0])
M_253  = 5.0e10;   c_253  = np.array([11.5, 0.0, 0.0])
M_247  = 3.0e9;    c_247  = np.array([0.0, 11.1, 0.0])

# 1. EXAKTER RECHNERISCHER PAARWEISER ABGLEICH (Die 9 Werte)
# Vektoren zwischen den Zentren berechnen
def calc_coupling_value(m1, m2, pos1, pos2):
    vec = pos2 - pos1
    dist = np.linalg.norm(vec)
    # Wenn Abstand gegen Null geht (Selbstwechselwirkung), setzen wir ihn auf Toleranz
    if dist < 1e-5:
        return 1.0, 0.0
    # Numerische Abrundung/Toleranzglättung für das asymmetrische Dreieck
    if abs(dist - 11.3) < 0.3:
        dist = np.round(dist / TOLERANZ) * TOLERANZ

    # Richtungs-Sinus näherungsweise über orthogonale Achsenlagen bestimmen
    sin_theta = np.sin(np.arctan2(vec[1], vec[0] + 1e-9))
    if sin_theta == 0: sin_theta = 1.0 # Sicherung der Kreuzspannung
    return dist, np.abs(sin_theta)

# Matrix-Berechnung für die Konsolenausgabe
objects = [("Sgr A*", M_SgrA, c_SgrA), ("NGC 253", M_253, c_253), ("NGC 247", M_247, c_247)]
coupling_matrix_vals = np.zeros((3, 3))

print("="*60)
print("   RVT COUPLING MATRIX VALUES (PROTOTYP-AUSGABE)")
print("="*60)
for i, (name1, m1, p1) in enumerate(objects):
    for j, (name2, m2, p2) in enumerate(objects):
        d, st = calc_coupling_value(m1, m2, p1, p2)
        # Kopplungswert über Massenprodukt und Sinus-Geometrie
        val = (m1 * m2) * st / (d**2 + 1e-5)
        coupling_matrix_vals[i, j] = val
        print(f"{name1} -> {name2}: Dist: {d:.3f} Mio Lj | RVT-Wert: {val:.3e}")
print("="*60 + "\n")


# 2. DETEKTOR-GITTER UND TRANSFORMATION (UMGEKEHRTE ENTFERNUNG)
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

def inverted_distance_field(center):
    rx = X - center[0]
    ry = Y - center[1]
    rz = Z - center[2]
    dist = np.sqrt(rx**2 + ry**2 + rz**2)
    # Toleranz einbauen, um Singularität im Kern zu verhindern
    dist = np.where(dist < TOLERANZ, TOLERANZ, dist)
    return 1.0 / dist # UMGEKEHRTE Entfernung für die haptische Speicherfunktion

# Invertierte Distanzfelder
inv_d_SgrA = inverted_distance_field(c_SgrA)
inv_d_253  = inverted_distance_field(c_253)
inv_d_247  = inverted_distance_field(c_247)

# Fusionierung der inversen Geometriefelder gewichtet mit den Kopplungswerten
# Wir nutzen die Diagonalelemente und Kreuzkopplungen der 9 berechneten Werte
Xi_inv = (coupling_matrix_vals[0,1] * inv_d_SgrA * inv_d_253 +
          coupling_matrix_vals[0,2] * inv_d_SgrA * inv_d_247 +
          coupling_matrix_vals[1,2] * inv_d_253 * inv_d_247)

# 3. DIFFERENTIELLER TENSOR-APPARAT (ALTE FUNKTION)
def second_derivative(f, dx, axis):
    first = np.gradient(f, dx, axis=axis, edge_order=2)
    return np.gradient(first, dx, axis=axis, edge_order=2)

d2x_Xi = second_derivative(Xi_inv, dx, axis=0)
d2y_Xi = second_derivative(Xi_inv, dx, axis=1)
d2z_Xi = second_derivative(Xi_inv, dx, axis=2)

laplace_Xi = d2x_Xi + d2y_Xi + d2z_Xi

E_xx = d2x_Xi - laplace_Xi
E_yy = d2y_Xi - laplace_Xi
E_zz = d2z_Xi - laplace_Xi

S_2D = E_xx + E_yy
M_RVT_final = S_2D - E_zz

# Schnitt bei z = 0 extrahieren
z_slice = N // 2
M_RVT_slice = M_RVT_final[:, :, z_slice]

# --- VISUALISIERUNG DER SPEICHERFUNKTION MIT GEKEHRTER METRIK ---
plt.figure(figsize=(10, 8))
plt.imshow(M_RVT_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='inferno', aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'orange'], s=100, edgecolors='white')

# Kennzeichnung des Toleranzkorridors der Hypotenuse
plt.plot([c_253[0], c_247[0]], [c_253[1], c_247[1]], color='green', linestyle='--', alpha=0.7, label='Hypotenuse (Toleranz-stabilisiert)')

plt.colorbar(label=r'Invertierte RVT-Speicherspannung $\mathcal{M}_{\mathrm{RVT}}$')
plt.title('RVT-Speicherfunktion (Umgekehrte Distanzmetrik)\nSgr A* Ursprung | Hypotenusen-Toleranz: 26.000 Lj', fontsize=12, fontweight='bold')
plt.xlabel('x [Mio. Lj]'); plt.ylabel('y [Mio. Lj]')
plt.xlim([-3, L/2]); plt.ylim([-3, L/2])
plt.legend()
plt.grid(True, alpha=0.15)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ============================================================
# RVT TRILATERALE MATRIX & INVERSE TENSOR-PROJEKTION
# Toleranz: 26.000 Lichtjahre (0.026 Mio. Lj) gegen Asymmetrie
# ============================================================

# --- Parameter ---
N = 140          # Gitterauflösung
L = 40.0         # Bereich [-20, +20] Millionen Lichtjahre
C_RVT = 0.8      # Kopplungskonstante
TOLERANZ = 0.026 # 26.000 Lj Toleranzgrenze (Mio. Lj)

# --- Massen & Koordinaten (Mio. Lj & M_sun) ---
M_SgrA = 4.154e6;   c_SgrA = np.array([0.0, 0.0, 0.0])
M_253  = 5.0e10;    c_253  = np.array([11.5, 0.0, 0.0])
M_247  = 3.0e9;     c_247  = np.array([0.0, 11.1, 0.0])

# Neigungswinkel für die Sinus-Modulation
inc_SgrA = np.radians(90.0)
inc_253  = np.radians(78.0)
inc_247  = np.radians(74.0)

# --- SCHRITT 1: EXPLIZITE BERECHNUNG DER 9 KOPPLUNGSWERTE ---
objects = [
    ("Sgr A*", M_SgrA, c_SgrA, inc_SgrA),
    ("NGC 253", M_253, c_253, inc_253),
    ("NGC 247", M_247, c_247, inc_247)
]

coupling_matrix = np.zeros((3, 3))

print("="*70)
print("   ERGEBNIS DER 9 PAARWEISEN KOPPLUNGSWERTE (SINUS-ABGLEICH)")
print("="*70)

for i, (name1, m1, p1, inc1) in enumerate(objects):
    for j, (name2, m2, p2, inc2) in enumerate(objects):
        if i == j:
            # Selbstwechselwirkung im Nullpunkt mit Toleranz abgesichert
            dist = TOLERANZ
            sin_mod = np.sin(inc1)
        else:
            # Reale Entfernung zwischen den Objekten
            raw_dist = np.linalg.norm(p2 - p1)
            # Toleranzabgleich für die Hypotenuse / asymmetrische Katheten
            if abs(raw_dist - 16.0) < 1.0 or abs(raw_dist - 11.3) < 1.0:
                dist = np.round(raw_dist / TOLERANZ) * TOLERANZ
            else:
                dist = raw_dist
            # Sinus-Kopplungskomponente aus beiden Winkeln
            sin_mod = np.sin(inc1) * np.sin(inc2)

        # Wertberechnung: Massenverknüpfung geteilt durch Geometrie-Abstand
        val = (m1 * m2) * sin_mod / (dist**2)
        coupling_matrix[i, j] = val
        print(f"[{i+1},{j+1}] {name1:8} -> {name2:8} | Dist: {dist:6.3f} Mio Lj | Wert: {val:.3e}")

print("="*70 + "\n")

# --- SCHRITT 2: ENTFTERNUING INVERTIEREN & 3D-GITTER AUFBAUEN ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# Funktion für das umgekehrte Distanzfeld (1/d)
def get_inverted_field(center):
    r = np.sqrt((X - center[0])**2 + (Y - center[1])**2 + (Z - center[2])**2)
    # Singulärer Kernschutz durch deine 26.000 Lj Toleranz
    r = np.where(r < TOLERANZ, TOLERANZ, r)
    return 1.0 / r

inv_F_SgrA = get_inverted_field(c_SgrA)
inv_F_253  = get_inverted_field(c_253)
inv_F_247  = get_inverted_field(c_247)

# Kombiniertes Feld gesteuert durch die berechneten Koppelkoeffizienten
# (Massen bleiben strukturell erhalten, Entfernungen wirken reziprok)
Xi_inv = (coupling_matrix[0, 1] * inv_F_SgrA * inv_F_253 +
          coupling_matrix[0, 2] * inv_F_SgrA * inv_F_247 +
          coupling_matrix[1, 2] * inv_F_253 * inv_F_247)

# --- SCHRITT 3: DIE URSPRÜNGLICHE TENSORFUNKTION (Zweite Ableitungen) ---
def second_deriv(f, delta, axis_idx):
    g1 = np.gradient(f, delta, axis=axis_idx, edge_order=2)
    return np.gradient(g1, delta, axis=axis_idx, edge_order=2)

d2x = second_deriv(Xi_inv, dx, 0)
d2y = second_deriv(Xi_inv, dx, 1)
d2z = second_deriv(Xi_inv, dx, 2)

laplace = d2x + d2y + d2z

# Korrekturtensor E_mu_nu
E_xx = d2x - laplace
E_yy = d2y - laplace
E_zz = d2z - laplace

# Projektiertes RVT-Speicherfeld M
S_2D = E_xx + E_yy
M_RVT = S_2D - E_zz

# Slicing bei z = 0
z_idx = N // 2
M_RVT_slice = M_RVT[:, :, z_idx]
S_2D_slice = S_2D[:, :, z_idx]
Xi_field_slice = Xi_inv[:, :, z_idx]

# --- SCHRITT 4: VISUALISIERUNG MIT OPTIMIERTER FARBDYNAMIK ---
plt.figure(figsize=(18, 5))

# Plot 1: Das reziproke Inversionsfeld Xi (Starke Farbdynamik: magma)
plt.subplot(1, 3, 1)
plt.imshow(np.log10(np.abs(Xi_field_slice) + 1e-5).T, origin='lower',
           extent=[-L/2, L/2, -L/2, L/2], cmap='magma', aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label=r'Log10 Inverse Feldstärke $\Xi_{\mathrm{inv}}$')
plt.title('1. Invertiertes Quellfeld (Magma-Dynamik)')
plt.xlabel('x [Mio. Lj]'); plt.ylabel('y [Mio. Lj]')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

# Plot 2: 2D-Projektionsspannung S_2D (Dynamik: coolwarm)
plt.subplot(1, 3, 2)
plt.imshow(S_2D_slice.T, origin='lower', extent=[-L/2, L/2, -L/2, L/2], cmap='coolwarm', aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label=r'$\mathcal{S}_{2D}$')
plt.title('2. 2D-Projektionsspannung Tensor')
plt.xlabel('x [Mio. Lj]')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

# Plot 3: Finale RVT-Speicherfunktion (Dynamik: seismic für Nullstellendifferenz)
plt.subplot(1, 3, 3)
# Symmetrisches Limit für seismic, um Nullpunkte perfekt darzustellen
v_max = np.max(np.abs(M_RVT_slice)) * 0.01  # Skalierung für feine Strukturen
plt.imshow(M_RVT_slice.T, origin='lower', extent=[-L/2, L/2, -L/2, L/2],
           cmap='seismic', vmin=-v_max, vmax=v_max, aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label=r'$\mathcal{M}_{\mathrm{RVT}}$')
plt.title('3. RVT-Speicherfunktion (Seismic-Nullstellen)')
plt.xlabel('x [Mio. Lj]')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ============================================================
# RVT INTEGRAL-MODELL: DOPPEL-3D-ANALYSE (MAXIMIERTE DARSTELLUNG)
# Sgr A* Ursprung | Toleranzkorrektur: 26.000 Lj (0.026 Mio. Lj)
# ============================================================

# --- Parameter ---
N = 140          # Gitterpunkte für drakonische Präzision
L = 40.0         # Systemgröße in Millionen Lichtjahren [-20, +20]
C_RVT = 0.8      # RVT-Kopplungskonstante
TOLERANZ = 0.026 # 26.000 Lichtjahre Toleranzgrenze (Mio. Lj)

# --- Realer Wertekanon (Mio. Lj & M_sun) ---
M_SgrA = 4.154e6;   c_SgrA = np.array([0.0, 0.0, 0.0])   # Zentraler Anker
M_253  = 5.0e10;    c_253  = np.array([11.5, 0.0, 0.0])  # X-Achse
M_247  = 3.0e9;     c_247  = np.array([0.0, 11.1, 0.0])  # Y-Achse

inc_SgrA = np.radians(90.0)
inc_253  = np.radians(78.0)
inc_247  = np.radians(74.0)

# --- Gitter erstellen ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# 3D-Gradientenfunktion für den Sinus-Abgleich
def gradient_3d(f, delta_x):
    gx, gy, gz = np.gradient(f, delta_x, delta_x, delta_x, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

def cross_variance_sin(g1, g2):
    mag1 = np.linalg.norm(g1, axis=-1) + 1e-30
    mag2 = np.linalg.norm(g2, axis=-1) + 1e-30
    dot = np.sum(g1 * g2, axis=-1)
    cos_theta = dot / (mag1 * mag2)
    sin_theta = np.sqrt(1.0 - np.clip(cos_theta**2, 0.0, 1.0))
    return 2.0 * C_RVT * mag1 * mag2 * sin_theta


# ============================================================
# PHASE 1: URSPRÜNGLICHE TRILATERALE VARIANZ 3D (GAUSS-ÄTHER)
# ============================================================
scale_SgrA, scale_253, scale_247 = 0.5, 2.5, 1.2
gamma_SgrA = 1.0 + C_RVT * np.sin(inc_SgrA)

def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_SgrA = gaussian_field(M_SgrA, scale_SgrA, c_SgrA) * gamma_SgrA
Xi_253  = gaussian_field(M_253, scale_253, c_253) * np.sin(inc_253)
Xi_247  = gaussian_field(M_247, scale_247, c_247) * np.sin(inc_247)

V_orig_SgrA_253 = cross_variance_sin(gradient_3d(Xi_SgrA, dx), gradient_3d(Xi_253, dx))
V_orig_SgrA_247 = cross_variance_sin(gradient_3d(Xi_SgrA, dx), gradient_3d(Xi_247, dx))
V_orig_253_247  = cross_variance_sin(gradient_3d(Xi_253, dx), gradient_3d(Xi_247, dx))
V_orig_total    = V_orig_SgrA_253 + V_orig_SgrA_247 + V_orig_253_247

# Plot 1: MAXIMIERTE 3D-Ansicht (Gauß)
fig1 = plt.figure(figsize=(12, 9))
ax1 = fig1.add_subplot(111, projection='3d')
for z_fix in [N // 4, N // 2, 3 * N // 4]:
    ax1.contour(X[:, :, z_fix], Y[:, :, z_fix], V_orig_total[:, :, z_fix], levels=12, cmap='viridis', alpha=0.5, zdir='z', offset=z[z_fix])

ax1.scatter(0, 0, 0, color='magenta', s=150, marker='o', label='Sgr A* (Anker)')
ax1.scatter(c_253[0], c_253[1], c_253[2], color='cyan', s=120, marker='x', label='NGC 253')
ax1.scatter(c_247[0], c_247[1], c_247[2], color='orange', s=120, marker='x', label='NGC 247')
ax1.set_xlim(-5, 15); ax1.set_ylim(-5, 15); ax1.set_zlim(-5, 5)
ax1.set_xlabel('x [Mio. Lj]'); ax1.set_ylabel('y [Mio. Lj]'); ax1.set_zlabel('z [Mio. Lj]')
ax1.set_title('1. MAXIMIERTE TRILATERALE 3D-VARIANZ (Gauß-Modell)', fontsize=14, fontweight='bold')
ax1.legend(loc='upper left')
plt.show()


# ============================================================
# PHASE 2: BERECHNUNG UND SPEICHERUNG DER 9 MATRIXWERTE
# ============================================================
objects = [("Sgr A*", M_SgrA, c_SgrA, inc_SgrA), ("NGC 253", M_253, c_253, inc_253), ("NGC 247", M_247, c_247, inc_247)]
coupling_matrix = np.zeros((3, 3))

print("\n" + "="*75)
print("   SPEICHER-LOGBUCH: DIE 9 DISKRETEN RVT-KOPPLUNGSWERTE (SINUS-ABGLEICH)")
print("="*75)
for i, (name1, m1, p1, in1) in enumerate(objects):
    for j, (name2, m2, p2, in2) in enumerate(objects):
        if i == j:
            dist = TOLERANZ
            sin_mod = np.sin(in1)
        else:
            raw_dist = np.linalg.norm(p2 - p1)
            if abs(raw_dist - 16.0) < 1.0 or abs(raw_dist - 11.3) < 1.0:
                dist = np.round(raw_dist / TOLERANZ) * TOLERANZ
            else:
                dist = raw_dist
            sin_mod = np.sin(in1) * np.sin(in2)

        val = (m1 * m2) * sin_mod / (dist**2)
        coupling_matrix[i, j] = val
        print(f"Index [{i+1},{j+1}] | {name1:8} -> {name2:8} | Korrigierte Dist: {dist:6.3f} Mio Lj | RVT-Wert: {val:.3e}")
print("="*75 + "\n")


# ============================================================
# PHASE 3: INVERTIERTE DISTANZMETREIK & ZWEITE 3D-VARIANZ
# ============================================================
def get_inverted_field(center):
    r = np.sqrt((X - center[0])**2 + (Y - center[1])**2 + (Z - center[2])**2)
    r = np.where(r < TOLERANZ, TOLERANZ, r)
    return 1.0 / r

inv_F_SgrA = get_inverted_field(c_SgrA)
inv_F_253  = get_inverted_field(c_253)
inv_F_247  = get_inverted_field(c_247)

Xi_inv = (coupling_matrix[0, 1] * inv_F_SgrA * inv_F_253 +
          coupling_matrix[0, 2] * inv_F_SgrA * inv_F_247 +
          coupling_matrix[1, 2] * inv_F_253 * inv_F_247)

V_inv_SgrA_253 = cross_variance_sin(gradient_3d(coupling_matrix[0,1]*inv_F_SgrA, dx), gradient_3d(inv_F_253, dx))
V_inv_SgrA_247 = cross_variance_sin(gradient_3d(coupling_matrix[0,2]*inv_F_SgrA, dx), gradient_3d(inv_F_247, dx))
V_inv_253_247  = cross_variance_sin(gradient_3d(inv_F_253, dx), gradient_3d(inv_F_247, dx))
V_inv_total    = V_inv_SgrA_253 + V_inv_SgrA_247 + V_inv_253_247

# Plot 2: MAXIMIERTE 3D-Ansicht (Invertierte Geometrie)
fig2 = plt.figure(figsize=(12, 9))
ax2 = fig2.add_subplot(111, projection='3d')
for z_fix in [N // 4, N // 2, 3 * N // 4]:
    # Logarithmierte Stufen für das reziproke Feld, da 1/d extreme Spitzen erzeugt
    ax2.contour(X[:, :, z_fix], Y[:, :, z_fix], np.log10(V_inv_total[:, :, z_fix] + 1e-5), levels=12, cmap='plasma', alpha=0.5, zdir='z', offset=z[z_fix])

ax2.scatter(0, 0, 0, color='magenta', s=150, marker='o', label='Sgr A* (Anker)')
ax2.scatter(c_253[0], c_253[1], c_253[2], color='cyan', s=120, marker='x', label='NGC 253')
ax2.scatter(c_247[0], c_247[1], c_247[2], color='orange', s=120, marker='x', label='NGC 247')
ax2.set_xlim(-3, 15); ax2.set_ylim(-3, 15); ax2.set_zlim(-5, 5)
ax2.set_xlabel('x [Mio. Lj]'); ax2.set_ylabel('y [Mio. Lj]'); ax2.set_zlabel('z [Mio. Lj]')
ax2.set_title('3. MAXIMIERTE INVERTIERTE TRILATERALE 3D-VARIANZ (1/d Metrik)', fontsize=14, fontweight='bold')
ax2.legend(loc='upper left')
plt.show()


# ============================================================
# PHASE 4: TENSORFUNKTION & OPTIMIERTE 2D-PROJEKTIONEN (BEREINIGT)
# ============================================================
def second_deriv(f, delta, axis_idx):
    g1 = np.gradient(f, delta, axis=axis_idx, edge_order=2)
    return np.gradient(g1, delta, axis=axis_idx, edge_order=2)

d2x = second_deriv(Xi_inv, dx, 0)
d2y = second_deriv(Xi_inv, dx, 1)
d2z = second_deriv(Xi_inv, dx, 2)
laplace = d2x + d2y + d2z

E_xx = d2x - laplace; E_yy = d2y - laplace; E_zz = d2z - laplace
S_2D = E_xx + E_yy
M_RVT = S_2D - E_zz

z_idx = N // 2
M_RVT_slice = M_RVT[:, :, z_idx]
S_2D_slice = S_2D[:, :, z_idx]
Xi_field_slice = Xi_inv[:, :, z_idx]

# 2D-Plots ausführen mit korrekten Raw-Strings (r'...') gegen SyntaxWarnings
plt.figure(figsize=(18, 5))
plt.subplot(1, 3, 1)
plt.imshow(np.log10(np.abs(Xi_field_slice) + 1e-5).T, origin='lower', extent=[-L/2, L/2, -L/2, L/2], cmap='magma', aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label='Log10 Field')
plt.title(r'Quellfeld $\Xi_{\mathrm{inv}}$ (Magma)')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

plt.subplot(1, 3, 2)
plt.imshow(S_2D_slice.T, origin='lower', extent=[-L/2, L/2, -L/2, L/2], cmap='coolwarm', aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label='S_2D')
plt.title(r'Projektionsspannung $\mathcal{S}_{2D}$ ')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

plt.subplot(1, 3, 3)
v_max = np.max(np.abs(M_RVT_slice)) * 0.01
plt.imshow(M_RVT_slice.T, origin='lower', extent=[-L/2, L/2, -L/2, L/2], cmap='seismic', vmin=-v_max, vmax=v_max, aspect='equal')
plt.scatter([0, c_253[0], c_247[0]], [0, c_253[1], c_247[1]], c=['magenta', 'cyan', 'yellow'], s=80, edgecolors='black')
plt.colorbar(label='M_RVT')
plt.title(r'Speicherfunktion $\mathcal{M}_{\mathrm{RVT}}$ (Seismic)')
plt.xlim([-3, 15]); plt.ylim([-3, 15])

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from skimage.measure import marching_cubes
from scipy.ndimage import gaussian_filter
import warnings
warnings.filterwarnings("ignore")

# ==============================================================================
# RVT TRILATERALE GEFÜGE-MATRIX MIT DEKLINATION & SANFTEM 3D-VOLUMENPLOTTING
# ==============================================================================

# --- Parameter ---
N_anim = 60      # Höhere Auflösung für sanfte 3D-Flächenstrukturen
L = 40.0         # Systemgröße in Millionen Lichtjahren [-20, +20]
dx = L / (N_anim - 1)
C_RVT = 0.8      # RVT-Kopplungskonstante
TOLERANZ = 0.026 # 26.000 Lj Toleranzgrenze gegen Asymmetrie

# --- Realer Wertekanon mit Sgr A* Deklinations-Erweiterung ---
# 1. Sgr A* (Zentraler Anker mit realer Deklination)
M_SgrA = 4.154e6
scale_SgrA = 0.8
center_SgrA = np.array([0.0, 0.0, 0.0])
declination_SgrA_deg = -29.008  # Reale Deklination von Sgr A*
inclination_SgrA_deg = 90.0

# 2. NGC 253 (X-Achsen Starburst)
M_253 = 5.0e10
scale_253 = 2.5
center_253 = np.array([11.5, 0.0, 0.0])
inclination_253_deg = 78.0

# 3. NGC 247 (Y-Achsen Partner)
M_247 = 3.0e9
scale_247 = 1.2
center_247 = np.array([0.0, 11.1, 0.0])
inclination_247_deg = 74.0

# Winkelumrechnungen in Radiant
inc_SgrA = np.radians(inclination_SgrA_deg)
dec_SgrA = np.radians(declination_SgrA_deg)
inc_253  = np.radians(inclination_253_deg)
inc_247  = np.radians(inclination_247_deg)

# --- RVT-Gamma-Modulation inklusive Sgr A* Deklinations-Vektor ---
# Sgr A* koppelt nun trilateral über Inklination und Deklination
gamma_SgrA = 1.0 + C_RVT * np.sin(inc_SgrA) * np.sin(dec_SgrA)
gamma_253  = 1.0 + C_RVT * np.sin(inc_253)
gamma_247  = 1.0 + C_RVT * np.sin(inc_247)

# --- Gitter erstellen ---
x = np.linspace(-L/2, L/2, N_anim)
y = np.linspace(-L/2, L/2, N_anim)
z = np.linspace(-L/2, L/2, N_anim)
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Ausrechnen und Ausgeben der 9 diskreten Kopplungswerte ---
objects = [
    ("Sgr A*", M_SgrA, center_SgrA, inc_SgrA, dec_SgrA),
    ("NGC 253", M_253, center_253, inc_253, 0.0),
    ("NGC 247", M_247, center_247, inc_247, 0.0)
]

coupling_matrix = np.zeros((3, 3))

print("\n" + "="*80)
print("   SPEICHER-LOGBUCH: DIE 9 KOPPLUNGSWERTE MIT SGR A* DEKLINATION")
print("="*80)
for i, (name1, m1, p1, in1, de1) in enumerate(objects):
    for j, (name2, m2, p2, in2, de2) in enumerate(objects):
        if i == j:
            dist = TOLERANZ
            # Eigensinus-Kopplung
            sin_mod = np.sin(in1) if i != 0 else np.sin(in1) * np.sin(de1)
        else:
            raw_dist = np.linalg.norm(p2 - p1)
            # Hypotenusen-Toleranzglättung
            if abs(raw_dist - 16.0) < 1.0 or abs(raw_dist - 11.3) < 1.0:
                dist = np.round(raw_dist / TOLERANZ) * TOLERANZ
            else:
                dist = raw_dist

            # Kreuzspannungs-Sinus
            if i == 0 or j == 0:
                sin_mod = np.sin(in1) * np.sin(in2) * np.sin(dec_SgrA)
            else:
                sin_mod = np.sin(in1) * np.sin(in2)

        val = (m1 * m2) * np.abs(sin_mod) / (dist**2)
        coupling_matrix[i, j] = val
        print(f"Knoten [{i+1},{j+1}] | {name1:8} -> {name2:8} | Effektive Dist: {dist:6.3f} Mio Lj | RVT-Wert: {val:.3e}")
print("="*80 + "\n")


# --- Generierung der 3 Gauß-Projektionsfelder ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_SgrA_base = gaussian_filter(gaussian_field(M_SgrA, scale_SgrA, center_SgrA), sigma=1.0) * gamma_SgrA
Xi_253_base  = gaussian_filter(gaussian_field(M_253, scale_253, center_253), sigma=1.0) * np.sin(inc_253)
Xi_247_base  = gaussian_filter(gaussian_field(M_247, scale_247, center_247), sigma=1.0) * np.sin(inc_247)

# --- Gradienten und trilaterale Feldeffekt-Varianz V_12 (Gauß-Ebenen) ---
def gradient_3d(f, delta_x):
    gx, gy, gz = np.gradient(f, delta_x, delta_x, delta_x, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

grad_SgrA = gradient_3d(Xi_SgrA_base, dx)
grad_253  = gradient_3d(Xi_253_base, dx)
grad_247  = gradient_3d(Xi_247_base, dx)

# Paarweise Kreuzkomponenten (Sinus-Kopplung über Skalarprodukt-Identität)
def cross_variance_sin(g1, g2):
    mag1 = np.linalg.norm(g1, axis=-1) + 1e-30
    mag2 = np.linalg.norm(g2, axis=-1) + 1e-30
    dot = np.sum(g1 * g2, axis=-1)
    cos_theta = dot / (mag1 * mag2)
    sin_theta = np.sqrt(1.0 - np.clip(cos_theta**2, 0.0, 1.0))
    return 2.0 * C_RVT * mag1 * mag2 * sin_theta

V_SgrA_253 = cross_variance_sin(grad_SgrA, grad_253)
V_SgrA_247 = cross_variance_sin(grad_SgrA, grad_247)
V_253_247  = cross_variance_sin(grad_253, grad_247)

# Totale trilaterale Varianz des stetigen Gefüges
V_12_total = V_SgrA_253 + V_SgrA_247 + V_253_247

# Negentropie-Vektor für die räumliche Strömungsrichtung
S_vector = -gradient_3d(V_12_total, dx)


# ==============================================================================
# VISUALISIERUNG: GROSSES 3D-GEFÜGE UND TENSOR-PROJEKTION
# ==============================================================================
fig = plt.figure(figsize=(16, 12))
ax_3d = fig.add_subplot(121, projection='3d')
ax_2d = fig.add_subplot(122)

# --- LINKER PLOT: SANFT GEFÄRBTES 3D-VOLUMENGEFÜGE ---
ax_3d.scatter(0, 0, 0, color='magenta', s=220, marker='o', edgecolors='black', label='Sgr A* Anker')
ax_3d.scatter(center_253[0], center_253[1], center_253[2], color='cyan', s=180, marker='x', linewidths=3, label='NGC 253')
ax_3d.scatter(center_247[0], center_247[1], center_247[2], color='orange', s=180, marker='x', linewidths=3, label='NGC 247')

# Generierung der sanften Isomesh-Flächen der trilateralen Koppelspannung
try:
    # Schwellenwert auf 8% des Maximums für feine, sanfte volumetrische Brückenstrukturen
    threshold = 0.08 * np.max(V_12_total)
    if threshold > 0:
        verts, faces, _, _ = marching_cubes(V_12_total, threshold, spacing=(dx, dx, dx))
        # Zentrierung des Meshs auf den Koordinatenraum [-L/2, L/2]
        verts_scaled = verts - [L/2, L/2, L/2]

        # Poly3DCollection für sanfte Farbdynamik (coolwarm/viridis-Übergang)
        mesh = Poly3DCollection(verts_scaled[faces], alpha=0.25, cmap='coolwarm', edgecolor='none')
        mesh.set_array(verts_scaled[:, 0]) # Sanfte Einfärbung entlang der x-Kopplungsachse
        ax_3d.add_collection3d(mesh)
except (ValueError, RuntimeError):
    pass

# Dünne Negentropie-Vektoren (Quiver) für den haptischen Fluss
skip = 6
ax_3d.quiver(X[::skip, ::skip, ::skip], Y[::skip, ::skip, ::skip], Z[::skip, ::skip, ::skip],
             S_vector[::skip, ::skip, ::skip, 0], S_vector[::skip, ::skip, ::skip, 1], S_vector[::skip, ::skip, ::skip, 2],
             length=1.5, normalize=True, color='purple', alpha=0.4, label=r'Negentropie $\vec{S}$')

ax_3d.set_xlim(-5, 15); ax_3d.set_ylim(-5, 15); ax_3d.set_zlim(-5, 5)
ax_3d.set_xlabel('x [Mio. Lj]'); ax_3d.set_ylabel('y [Mio. Lj]'); ax_3d.set_zlabel('z [Mio. Lj]')
ax_3d.set_title(r'3D: Sanftes Trilaterales Varianzgefüge $\mathcal{V}_{12}$', fontsize=12, fontweight='bold')
ax_3d.legend(loc='upper left')
ax_3d.view_init(elev=25, azim=-45)


# --- RECHTER PLOT: INFRASTRUKTUR DER 2D-PROJEKTIONSPANNUNG (S_2D) ---
# Zurückgreifen auf die d2x/d2y-Gefügespannung aus dem vorherigen Tensor
laplace_Xi = (np.roll(Xi_12 := Xi_SgrA_base + Xi_253_base + Xi_247_base, 1, axis=0) + np.roll(Xi_12, -1, axis=0) +
              np.roll(Xi_12, 1, axis=1) + np.roll(Xi_12, -1, axis=1) +
              np.roll(Xi_12, 1, axis=2) + np.roll(Xi_12, -1, axis=2) - 6 * Xi_12) / (dx**2)

E_xx = (np.roll(Xi_12, 1, axis=0) + np.roll(Xi_12, -1, axis=0) - 2 * Xi_12) / (dx**2) - laplace_Xi
E_yy = (np.roll(Xi_12, 1, axis=1) + np.roll(Xi_12, -1, axis=1) - 2 * Xi_12) / (dx**2) - laplace_Xi
S_2D_tensor = gaussian_filter(E_xx + E_yy, sigma=1.0)

z_slice = N_anim // 2
S_2D_slice = S_2D_tensor[:, :, z_slice]

contour_2d = ax_2d.contourf(X[:, :, z_slice], Y[:, :, z_slice], S_2D_slice, levels=25, cmap='plasma', alpha=0.8)
fig.colorbar(contour_2d, ax=ax_2d, label=r'$\mathcal{S}_{2D}$')
ax_2d.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=120, edgecolors='black')

ax_2d.set_xlim(-3, 15); ax_2d.set_ylim(-3, 15)
ax_2d.set_xlabel('x [Mio. Lj]'); ax_2d.set_ylabel('y [Mio. Lj]')
ax_2d.set_title(r'2D-Projektion: $\mathcal{S}_{2D}$ Feldspannung', fontsize=12, fontweight='bold')
ax_2d.grid(True, alpha=0.15)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from skimage.measure import marching_cubes
from scipy.ndimage import gaussian_filter
import warnings
warnings.filterwarnings("ignore")

# ==============================================================================
# RVT TRILATERALE MATRIX: REIN RECHTWINKLIGES GEFÜGE (4 SPEZIFISCHE SICHTACHSEN)
# ==============================================================================

# --- Parameter ---
N_anim = 60      # Höhere Auflösung für sanfte 3D-Flächenstrukturen
L = 40.0         # Systemgröße in Millionen Lichtjahren [-20, +20]
dx = L / (N_anim - 1)
C_RVT = 0.8      # RVT-Kopplungskonstante
TOLERANZ = 0.026 # 26.000 Lj Toleranzgrenze gegen Asymmetrie

# --- Realer Wertekanon (OHNE Deklination) ---
# 1. Sgr A* (Zentraler Anker - reines Basissystem)
M_SgrA = 4.154e6
scale_SgrA = 0.8
center_SgrA = np.array([0.0, 0.0, 0.0])
inclination_SgrA_deg = 90.0

# 2. NGC 253 (X-Achsen Starburst)
M_253 = 5.0e10
scale_253 = 2.5
center_253 = np.array([11.5, 0.0, 0.0])
inclination_253_deg = 78.0

# 3. NGC 247 (Y-Achsen Partner)
M_247 = 3.0e9
scale_247 = 1.2
center_247 = np.array([0.0, 11.1, 0.0])
inclination_247_deg = 74.0

# Winkelumrechnungen in Radiant
inc_SgrA = np.radians(inclination_SgrA_deg)
inc_253  = np.radians(inclination_253_deg)
inc_247  = np.radians(inclination_247_deg)

# --- RVT-Gamma-Modulation (OHNE Deklinations-Einfluss) ---
gamma_SgrA = 1.0 + C_RVT * np.sin(inc_SgrA)
gamma_253  = 1.0 + C_RVT * np.sin(inc_253)
gamma_247  = 1.0 + C_RVT * np.sin(inc_247)

# --- Gitter erstellen ---
x = np.linspace(-L/2, L/2, N_anim)
y = np.linspace(-L/2, L/2, N_anim)
z = np.linspace(-L/2, L/2, N_anim)
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Ausrechnen und Ausgeben der 9 diskreten Kopplungswerte ---
objects = [
    ("Sgr A*", M_SgrA, center_SgrA, inc_SgrA),
    ("NGC 253", M_253, center_253, inc_253),
    ("NGC 247", M_247, center_247, inc_247)
]

coupling_matrix = np.zeros((3, 3))

print("\n" + "="*80)
print("   SPEICHER-LOGBUCH: DIE 9 KOPPLUNGSWERTE (KLINISCH REINES DECK)")
print("="*80)
for i, (name1, m1, p1, in1) in enumerate(objects):
    for j, (name2, m2, p2, in2) in enumerate(objects):
        if i == j:
            dist = TOLERANZ
            sin_mod = np.sin(in1)
        else:
            raw_dist = np.linalg.norm(p2 - p1)
            if abs(raw_dist - 16.0) < 1.0 or abs(raw_dist - 11.3) < 1.0:
                dist = np.round(raw_dist / TOLERANZ) * TOLERANZ
            else:
                dist = raw_dist
            sin_mod = np.sin(in1) * np.sin(in2)

        val = (m1 * m2) * np.abs(sin_mod) / (dist**2)
        coupling_matrix[i, j] = val
        print(f"Knoten [{i+1},{j+1}] | {name1:8} -> {name2:8} | Effektive Dist: {dist:6.3f} Mio Lj | RVT-Wert: {val:.3e}")
print("="*80 + "\n")


# --- Generierung der 3 Gauß-Projektionsfelder (Fehler behoben!) ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_SgrA_base = gaussian_filter(gaussian_field(M_SgrA, scale_SgrA, center_SgrA), sigma=1.0) * gamma_SgrA
Xi_253_base  = gaussian_filter(gaussian_field(M_253, scale_253, center_253), sigma=1.0) * np.sin(inc_253)
Xi_247_base  = gaussian_filter(gaussian_field(M_247, scale_247, center_247), sigma=1.0) * np.sin(inc_247)

# --- Gradienten und trilaterale Feldeffekt-Varianz V_12 ---
def gradient_3d(f, delta_x):
    gx, gy, gz = np.gradient(f, delta_x, delta_x, delta_x, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

grad_SgrA = gradient_3d(Xi_SgrA_base, dx)
grad_253  = gradient_3d(Xi_253_base, dx)
grad_247  = gradient_3d(Xi_247_base, dx)

def cross_variance_sin(g1, g2):
    mag1 = np.linalg.norm(g1, axis=-1) + 1e-30
    mag2 = np.linalg.norm(g2, axis=-1) + 1e-30
    dot = np.sum(g1 * g2, axis=-1)
    cos_theta = dot / (mag1 * mag2)
    sin_theta = np.sqrt(1.0 - np.clip(cos_theta**2, 0.0, 1.0))
    return 2.0 * C_RVT * mag1 * mag2 * sin_theta

V_SgrA_253 = cross_variance_sin(grad_SgrA, grad_253)
V_SgrA_247 = cross_variance_sin(grad_SgrA, grad_247)
V_253_247  = cross_variance_sin(grad_253, grad_247)

V_12_total = V_SgrA_253 + V_SgrA_247 + V_253_247
S_vector = -gradient_3d(V_12_total, dx)

# Mesh extrahieren
threshold = 0.08 * np.max(V_12_total)
verts, faces, _, _ = marching_cubes(V_12_total, threshold, spacing=(dx, dx, dx))
verts_scaled = verts - [L/2, L/2, L/2]

# ==============================================================================
# VISUALISIERUNG: 4 SPEZIFISCH AUSGERICHTETE SICHTACHSEN (2x2)
# ==============================================================================
fig = plt.figure(figsize=(16, 14))

# Dimis exakte Wunsch-Perspektiven
views = [
    ("1. Isometrische Ansicht (45° Links / 25° Runter)", 25, -45),
    ("2. Ursprung-Seitenansicht (45° Hochgedreht)", 45, 0),
    ("3. Draufsicht (Reine X-Y Achse)", 90, -90),
    ("4. Schrägansicht (45° Rechts / 45° Runter)", -45, 45)
]

for idx, (title, elev, azim) in enumerate(views):
    ax = fig.add_subplot(2, 2, idx + 1, projection='3d')

    # Ankerpunkte einzeichnen
    ax.scatter(0, 0, 0, color='magenta', s=160, marker='o', edgecolors='black', label='Sgr A*' if idx==0 else "")
    ax.scatter(center_253[0], center_253[1], center_253[2], color='cyan', s=140, marker='x', linewidths=3, label='NGC 253' if idx==0 else "")
    ax.scatter(center_247[0], center_247[1], center_247[2], color='orange', s=140, marker='x', linewidths=3, label='NGC 247' if idx==0 else "")

    # Sanftes Volumenmesh
    mesh = Poly3DCollection(verts_scaled[faces], alpha=0.22, cmap='coolwarm', edgecolor='none')
    mesh.set_array(verts_scaled[:, 0])
    ax.add_collection3d(mesh)

    # Flussvektoren
    skip = 6
    ax.quiver(X[::skip, ::skip, ::skip], Y[::skip, ::skip, ::skip], Z[::skip, ::skip, ::skip],
              S_vector[::skip, ::skip, ::skip, 0], S_vector[::skip, ::skip, ::skip, 1], S_vector[::skip, ::skip, ::skip, 2],
              length=1.3, normalize=True, color='purple', alpha=0.25)

    ax.set_xlim(-4, 14); ax.set_ylim(-4, 14); ax.set_zlim(-5, 5)
    ax.set_xlabel('x [Mio. Lj]'); ax.set_ylabel('y [Mio. Lj]'); ax.set_zlabel('z [Mio. Lj]')
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.view_init(elev=elev, azim=azim)

    if idx == 0:
        ax.legend(loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ==============================================================================
# 1. RVT-BOOST: REINE SYSTEMANKER-TRANSFORMATION (Sgr A*, NGC 247, NGC 253)
# ==============================================================================

def rvt_boost(i_deg):
    i_rad = np.radians(i_deg)
    return 1 + 0.8 * np.sin(i_rad)

# --- Berechnung der korrigierten Real-Werte ---
M_scheinbar = {"SgrA": 4.1e6, "NGC253": 5.0e10, "NGC247": 3.0e9}
R_scheinbar = {"SgrA": 0.00001, "NGC253": 35000, "NGC247": 25000}  # Radien in Lj (UNVERÄNDERT!)

gamma_SgrA = rvt_boost(0)
gamma_247 = rvt_boost(74)
gamma_253 = rvt_boost(78)

# RVT-korrigierte Massen (Entfernungen bleiben gleich!)
M_tatsächlich_SgrA = M_scheinbar['SgrA'] * gamma_SgrA
M_tatsächlich_247 = M_scheinbar['NGC247'] * gamma_247
M_tatsächlich_253 = M_scheinbar['NGC253'] * gamma_253

print("\n=== RVT-korrigierte Massen (Entfernungen UNVERÄNDERT!) ===")
print(f"Sgr A*:  M_tatsächlich = {M_tatsächlich_SgrA:.2e} M☉ (Boost: {gamma_SgrA:.3f})")
print(f"NGC 247: M_tatsächlich = {M_tatsächlich_247:.2e} M☉ (Boost: {gamma_247:.3f})")
print(f"NGC 253: M_tatsächlich = {M_tatsächlich_253:.2e} M☉ (Boost: {gamma_253:.3f})")

# ==============================================================================
# 2. ROTATIONSKURVEN FÜR DIE REALE SYSTEMKASKADE (DEIN EXAKTER CODE!)
# ==============================================================================

def rotation_curve(R_tatsächlich):
    g_tatsächlich = 1.0
    return np.sqrt(g_tatsächlich * R_tatsächlich)

# Tatsächliche, dynamisch geboostete Radien
R_tat_247 = 25000 * 1.769  # NGC 247
R_tat_253 = 35000 * 1.783  # NGC 253

R = np.linspace(1, max(R_tat_247, R_tat_253), 200)

v_253 = rotation_curve(R) * np.sqrt(R_tat_253) / 100.0
v_247 = rotation_curve(R) * np.sqrt(R_tat_247) / 100.0

v_classic_253 = rotation_curve(R) * np.sqrt(35000) / 100.0
v_classic_247 = rotation_curve(R) * np.sqrt(25000) / 100.0

fig, ax = plt.subplots(figsize=(10, 6))
ax.plot(R, v_253, label=f'NGC 253 (RVT-korrigiert, $R_{{tatsächlich}}={R_tat_253:.0f}$ Lj)', color='cyan', linewidth=2)
ax.plot(R, v_247, label=f'NGC 247 (RVT-korrigiert, $R_{{tatsächlich}}={R_tat_247:.0f}$ Lj)', color='orange', linewidth=2)
ax.plot(R, v_classic_253, label='NGC 253 (klassisch, $R_{scheinbar}=35.000$ Lj)', linestyle='--', color='lightblue')
ax.plot(R, v_classic_247, label='NGC 247 (klassisch, $R_{scheinbar}=25.000$ Lj)', linestyle='--', color='bisque')

ax.set_xlabel('Radius $R$ (Lichtjahre)')
ax.set_ylabel('Rotationsgeschwindigkeit $v_{rot}$ (normiert)')
ax.set_title('RVT-korrigierte Rotationskurven im lokalen Tensorfeld')
ax.legend()
ax.grid(True)
plt.show()

# ============================================================
# 3. TRILATERALE COUPLING-MATRIX (SINUS-MODUS) + SKALARTENSORFELD VARIANZ 12
# ============================================================

# --- Parameter ---
N = 140
L = 40.0
C_RVT = 0.8

# --- Realer Wertekanon (Mio. Lj & M_sun) ---
M_SgrA = M_tatsächlich_SgrA
scale_SgrA = 0.5
center_SgrA = (0.0, 0.0, 0.0)

M_253 = M_tatsächlich_253
scale_253 = 2.5
center_253 = (11.5, 0.0, 0.0)
inclination_253_deg = 78.0

M_247 = M_tatsächlich_247
scale_247 = 1.2
center_247 = (0.0, 11.1, 0.0)
inclination_247_deg = 74.0

# Winkel in Radiant
inclination_253 = np.radians(inclination_253_deg)
inclination_247 = np.radians(inclination_247_deg)

# --- RVT-Gamma-Korrektur ---
gamma_SgrA = 1.0 + C_RVT * np.sin(np.radians(90.0))
gamma_253 = 1.0 + C_RVT * np.sin(inclination_253)
gamma_247 = 1.0 + C_RVT * np.sin(inclination_247)

# --- Gitter erstellen ---
x = np.linspace(-L / 2, L / 2, N)
y = np.linspace(-L / 2, L / 2, N)
z = np.linspace(-L / 2, L / 2, N)
dx = x[1] - x[0]
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# --- Funktion für zweite Ableitung ---
def second_derivative(f, dx, axis):
    first = np.gradient(f, dx, axis=axis, edge_order=2)
    return np.gradient(first, dx, axis=axis, edge_order=2)

# --- Gauß-Projektionsfelder ---
def gaussian_field(mass, scale, center):
    cx, cy, cz = center
    return mass * np.exp(-((X - cx)**2 + (Y - cy)**2 + (Z - cz)**2) / (2 * scale**2))

Xi_SgrA = gaussian_field(M_SgrA, scale_SgrA, center_SgrA) * gamma_SgrA
Xi_253 = gaussian_field(M_253, scale_253, center_253) * np.sin(inclination_253)
Xi_247 = gaussian_field(M_247, scale_247, center_247) * np.sin(inclination_247)

Xi = Xi_SgrA + Xi_253 + Xi_247

# --- 3D-Gradienten ---
def gradient_3d(f, dx):
    gx, gy, gz = np.gradient(f, dx, dx, dx, edge_order=2)
    return np.stack((gx, gy, gz), axis=-1)

grad_SgrA = gradient_3d(Xi_SgrA, dx)
grad_253 = gradient_3d(Xi_253, dx)
grad_247 = gradient_3d(Xi_247, dx)

# --- Trilaterale Sinus-Varianz ---
def cross_variance_sin(g1, g2):
    mag1 = np.linalg.norm(g1, axis=-1) + 1e-30
    mag2 = np.linalg.norm(g2, axis=-1) + 1e-30
    dot = np.sum(g1 * g2, axis=-1)
    cos_theta = dot / (mag1 * mag2)
    sin_theta = np.sqrt(1.0 - np.clip(cos_theta**2, 0.0, 1.0))
    return 2.0 * C_RVT * mag1 * mag2 * sin_theta

V_SgrA_253 = cross_variance_sin(grad_SgrA, grad_253)
V_SgrA_247 = cross_variance_sin(grad_SgrA, grad_247)
V_253_247 = cross_variance_sin(grad_253, grad_247)

V_12_total = V_SgrA_253 + V_SgrA_247 + V_253_247

# --- Skalartensorfeld ---
d2x_Xi = second_derivative(Xi, dx, axis=0)
d2y_Xi = second_derivative(Xi, dx, axis=1)
d2z_Xi = second_derivative(Xi, dx, axis=2)
laplace_Xi = d2x_Xi + d2y_Xi + d2z_Xi

E_xx = d2x_Xi - laplace_Xi
E_yy = d2y_Xi - laplace_Xi
E_zz = d2z_Xi - laplace_Xi

S_12 = E_xx + E_yy + E_zz

# --- Extraktion der 2D-Schnitte ---
z_slice = N // 2
V_12_slice = V_12_total[:, :, z_slice]
S_12_slice = S_12[:, :, z_slice]

print("\n=== SKALARTENSORFELD VARIANZ 12 (S_12) ===")
print(f"Maximalwert von S_12: {np.max(S_12_slice):.4e}")
print(f"Mittelwert von S_12: {np.mean(S_12_slice):.4e}")
print(f"Minimalwert von S_12: {np.min(S_12_slice):.4e}")

# --- Visualisierung ---
plt.figure(figsize=(18, 5))

plt.subplot(1, 3, 1)
plt.imshow(V_12_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='viridis', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=70, edgecolors='black')
plt.colorbar(label=r'$\mathcal{V}_{12}$')
plt.title(r'Trilaterale Sinus-Varianz $\mathcal{V}_{12}$')
plt.xlabel('x [Mio. Lj]'); plt.ylabel('y [Mio. Lj]')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

plt.subplot(1, 3, 2)
plt.imshow(S_12_slice.T, origin='lower', extent=[-L / 2, L / 2, -L / 2, L / 2], cmap='plasma', aspect='equal')
plt.scatter([0, center_253[0], center_247[0]], [0, center_253[1], center_247[1]], c=['magenta', 'cyan', 'orange'], s=70, edgecolors='black')
plt.colorbar(label=r'$\mathcal{S}_{12}$')
plt.title(r'Skalartensorfeld Varianz 12 $\mathcal{S}_{12}$')
plt.xlabel('x [Mio. Lj]')
plt.xlim([-5, L/2]); plt.ylim([-5, L/2])

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')
for z_fix in [N // 4, N // 2, 3 * N // 4]:
    ax.contour(X[:, :, z_fix], Y[:, :, z_fix], S_12[:, :, z_fix], levels=10, cmap='viridis', alpha=0.4, zdir='z', offset=z[z_fix])

ax.scatter(0, 0, 0, color='magenta', s=120, marker='o', label=f'Sgr A* ({M_SgrA:.2e} M☉)')
ax.scatter(center_253[0], center_253[1], center_253[2], color='cyan', s=80, marker='x', label=f'NGC 253 ({M_253:.2e} M☉)')
ax.scatter(center_247[0], center_247[1], center_247[2], color='orange', s=80, marker='x', label=f'NGC 247 ({M_247:.2e} M☉)')
ax.set_xlim(-5, L/2); ax.set_ylim(-5, L/2); ax.set_zlim(-L/4, L/4)
ax.set_xlabel('x [Mio. Lj]'); ax.set_ylabel('y [Mio. Lj]')
ax.set_title(r'3D-Skalartensorfeld Varianz 12')
ax.legend()
plt.show()
```
